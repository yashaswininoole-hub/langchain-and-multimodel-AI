### Image Embeddings

Vision models do NOT understand images directly like humans.

They convert images into numbers called:

Embeddings (vectors)

### Simple Idea

Cat Image → [0.23, 0.81, 0.45, ...]

Dog Image → [0.12, 0.77, 0.51, ...]

These numbers represent:

    * shapes
    * colors
    * patterns
    * objects
    * semantics

### CLIP is a model that learns:

Relationship between images and text

It understands:

    * images
    * captions
    * meanings
    
together.

### CLIP converts BOTH:

    images
    text

into embeddings.

### Internal Architecture

    Image → Image Encoder → Image Embedding

    Text → Text Encoder → Text Embedding

Then compares similarity.

### Why CLIP is Important

CLIP powers:

    * image search
    * multimodal AI
    * image captioning

    * MMRAG

```python
# CLIP + Cosine Similarity + FAISS
# Image Search Engine
```

```python
from sentence_transformers import SentenceTransformer
from PIL import Image
from sklearn.metrics.pairwise import cosine_similarity
import faiss
import numpy as np
```

```python
#  Load CLIP Model

model = SentenceTransformer('clip-ViT-B-32')
```

```python
#  Load Images

img1 = Image.open("dog1.jpg")
img2 = Image.open("dog2.jpg")
img3 = Image.open("car.jpeg")
```

```python
#  Create Embeddings

emb1 = model.encode(img1)
emb2 = model.encode(img2)
emb3 = model.encode(img3)
```

```python
#  Check Embedding Shape

print("Embedding Shape :", emb1.shape)
```

```python
# Cosine Similarity

score1 = cosine_similarity([emb1], [emb2])

score2 = cosine_similarity([emb1], [emb3])

print("\nDog1 vs Dog2 Similarity")
print(score1)

print("\nDog1 vs Car Similarity")
print(score2)
```

```python
#  Create FAISS Index

dimension = 512

index = faiss.IndexFlatL2(dimension)
```

```python
#  Store Embeddings in FAISS

vectors = np.array([
    emb1,
    emb2,
    emb3
]).astype('float32')

index.add(vectors)

print("\nTotal Images Stored :", index.ntotal)
```

```python
# Search Similar Images

query = np.array([emb1]).astype('float32')

D, I = index.search(query, k=3)
```

```python

#  Show Results
image_names = [
    "dog1.jpg",
    "dog2.jpg",
    "car.jpeg"
]
```

```python

print("\nMost Similar Images:\n")

for idx, distance in zip(I[0], D[0]):

    print("Image :", image_names[idx])
    print("Distance :", distance)
    print()
```

Smaller distance: more similar

Larger distance: less similar

### Text + Image Retrieval

```python
from sentence_transformers import SentenceTransformer
from PIL import Image
import faiss
import numpy as np
```

```python
# Load CLIP
model = SentenceTransformer('clip-ViT-B-32')

# Images
image_paths = [
    "dog1.jpg",
    "dog2.jpg",
    "car.jpeg"
]

images = [Image.open(path) for path in image_paths]
```

```python

# Normalized embeddings
image_embeddings = model.encode(
    images,
    normalize_embeddings=True
)

# FAISS cosine similarity index
index = faiss.IndexFlatIP(512)

# Add embeddings
index.add(
    np.array(image_embeddings).astype('float32')
)
```

```python

query = "a photo of a dog"
```

```python
# Text embedding
text_embedding = model.encode(
    [query],
    normalize_embeddings=True
)
```

```python
# Search
D, I = index.search(
    np.array(text_embedding).astype('float32'),
    k=2
)
```

```python

# Results
print("\nQuery :", query)

print("\nMost Relevant Images:\n")

for idx, score in zip(I[0], D[0]):

    print("Image :", image_paths[idx])
    print("Similarity Score :", score)
    print()
```

### Retrieval + Generation

```python
# MULTIMODAL RETRIEVAL + GENERATION
# CLIP + FAISS + LLaVA
```

```python

from sentence_transformers import SentenceTransformer
from PIL import Image
import faiss
import numpy as np
import ollama
```

```python
#  LOAD CLIP MODEL

clip_model = SentenceTransformer(
    'clip-ViT-B-32'
)
```

```python
#  IMAGE DATASET

image_paths = [
    "dog1.jpg",
    "dog2.jpg",
    "car.jpeg"
]
```

```python
#  LOAD IMAGES

images = []

for path in image_paths:

    img = Image.open(path)

    images.append(img)
```

```python
# IMAGE EMBEDDINGS

image_embeddings = clip_model.encode(
    images,
    normalize_embeddings=True
)
```

```python
#  CREATE FAISS INDEX

dimension = 512

index = faiss.IndexFlatIP(dimension)

index.add(
    np.array(image_embeddings).astype('float32')
)

print("Total Images :", index.ntotal)
```

```python
# USER QUERY

query = "a photo of a dog"


# TEXT EMBEDDING

text_embedding = clip_model.encode(
    [query],
    normalize_embeddings=True
)
```

```python
#  SEARCH SIMILAR IMAGES

D, I = index.search(
    np.array(text_embedding).astype('float32'),
    k=2
)
```

```python
#  RETRIEVED IMAGE

best_index = I[0][0]

best_image = image_paths[best_index]

print("\nRetrieved Image :")
print(best_image)

print("\nSimilarity Score :")
print(D[0][0])
```

```python
#  SEND IMAGE TO LLAVA

response = ollama.chat(
    model='llava:7b',
    messages=[
        {
            'role': 'user',
            'content':'Describe this image in detail',
            'images': [best_image]
        }
    ]
)
```

```python
print("\nLLaVA Explanation:\n")

print(
    response['message']['content']
)
```
