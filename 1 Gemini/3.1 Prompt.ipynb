
### Understanding User Prompts vs System Prompts



<b> A prompt is the instruction or input we give to an AI model to get a response. </b>


### The quality of the AI’s answer depends heavily on:

    * Clarity
    
    * Specificity
    
    * Context
    
    * Constraints


What is a <b>User Prompt</b>?

    A user prompt is the direct instruction or question you send to the AI model. 

    It's what the end user asks for.

What is a <b>System Prompt</b>?

    A system prompt (or system instruction) is a hidden instruction that sets the behavior, tone, and role of the AI.

    It shapes how the AI responds to user prompts.


A <b>User Instruction</b> is:

    The actual query
    
    The request from the user
    
    Dynamic input

It controls what the model should answer.


A<b> System Instruction </b>defines:

    The behavior
    
    The role
    
    The tone
    
    The rules
    
    The limitations

It is set by the developer and not visible to the end user.

It controls how the model behaves.


```python
from dotenv import load_dotenv

import os

from google import genai
```

```python
# Load .env
load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
```

```python
# Create Gemini client

client = genai.Client(api_key=api_key)
```

```python
from google import genai

from google.genai import types
```

```python

response = client.models.generate_content(
    
    model="gemini-2.5-flash",
    
    contents="Explain AI.",
    
    config=types.GenerateContentConfig(
    
        system_instruction="""
        You are a friendly AI teacher.
        Explain concepts in very simple language.
        Use short sentences.
        Give one real-life example.
        Keep the answer under 120 words.
        """
    )
)

print(response.text)
```

### Chain-of-Thought (CoT), where the model is encouraged to reason step-by-step before giving the final answer.

Why CoT is Useful

It helps the model:

    Break complex problems into smaller steps
    
    Improve logical reasoning
    
    Increase accuracy in tasks like math, coding, and reasoning.


```python
# User prompt
prompt = """
You are a careful math assistant.

Solve the problem using these steps:

1. Calculate the discount amount.
2. Calculate the price after discount.
3. Calculate the GST amount.
4. Calculate the final price.

Keep the explanation short.

Question:
A laptop costs ₹50,000.
A 10% discount is applied.
Then 18% GST is added.

Give the final answer clearly.
"""

# Generate response
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```
