## Explore generation parameters



| Parameter       | Controls             | Lower value | Higher value |
| --------------- | -------------------- | ----------- | ------------ |
| 🌡️ Temperature | Randomness           | Focused     | Creative     |
| 🔢 Top-K        | Number of candidates | Fewer       | More         |
| 🎯 Top-P        | Probability mass     | Narrower    | Broader      |


### Output length

✅ Output length affects both cost and performance.

⏳ More tokens = more computation, higher latency, more energy use, and higher cost.

⚙️ Use the <b>max_output_tokens</b> parameter in the Gemini API to set a limit on how many tokens the model can generate.

🚫 This parameter only stops generation after the limit — it doesn’t make the output shorter or more concise by itself.

🧠 To get a complete and meaningful answer within the limit, use prompt engineering (clear, specific, focused prompts).


### Temperature

Temperature controls how random or creative the model's output can be.
    
    🧊 Low temperature (e.g., 0.1–0.3) → focused, predictable, consistent answers
    
    ⚖️ Medium temperature (e.g., 0.5–0.7) → balanced creativity and consistency
    
    🔥 High temperature (e.g., 0.8–1.5) → more diverse, creative, and sometimes unpredictable answers
    
    ⚙️ Temperature = 0 → usually favors the most probable next token (greedy-like decoding)


|    Temperature | Typical use                              | Why?                                       |
| -------------: | ---------------------------------------- | ------------------------------------------ |
|    **0 – 0.2** | 📊 Data extraction, classification       | Very consistent output                     |
|    **0 – 0.3** | 💻 Coding / debugging                    | Less variation, more focused               |
|  **0.1 – 0.3** | 📚 Question answering / RAG              | Factual and consistent                     |
|  **0.3 – 0.5** | 🤖 Chatbots / customer support           | Natural but controlled                     |
|  **0.5 – 0.7** | ✍️ General writing                       | Good balance                               |
|  **0.7 – 1.0** | 📱 Social media content                  | More creative variations                   |
|  **0.8 – 1.2** | 📖 Story / creative writing              | More diversity and creativity              |
| **1.0 – 1.5+** | 🎨 Brainstorming / highly creative tasks | Very diverse, potentially less predictable |


🧠 Easy rule to remember

    Factual application → LOW temperature 🧊
    
    Creative application → HIGH temperature 🔥

### Top-K and top-P

🎛 Top-K and Top-P also control output diversity like temperature.

🔢 Top-K = number of most probable tokens to choose from.

        Example: topK = 1 → greedy decoding (always picks the top token).

🎯 Top-P = cumulative probability threshold for candidate tokens.

        Example: topP = 0 → greedy decoding; topP = 1 → includes all tokens.
        

```python
Assume a language model predicts the next word after:

“I love eating”

It outputs probabilities like this:
```

| Token    | Probability |
| -------- | ----------- |
| pizza    | 0.40        |
| biryani  | 0.25        |
| pasta    | 0.15        |
| icecream | 0.10        |
| salad    | 0.05        |
| sushi    | 0.03        |
| burger   | 0.02        |



Top-K = 1 : Always selects pizza

This is greedy decoding.


```python
Top-K = 3: 
            [pizza (0.40), biryani (0.25), pasta (0.15)]

            All others are removed.
            
            Now we renormalize probabilities:
            
            New total = 0.40 + 0.25 + 0.15 = 0.80

New probabilities:

            pizza → 0.40 / 0.80 = 0.50
            
            biryani → 0.25 / 0.80 = 0.31
            
            pasta → 0.15 / 0.80 = 0.19
            
            Now the model randomly picks among these 3.

👉 Output becomes more diverse than greedy.
```

Top-P (Nucleus Sampling)

Definition: Choose the smallest set of tokens whose cumulative probability ≥ P.

1. Top-P = 0.50

    Add tokens from highest probability until sum ≥ 0.50:

        pizza = 0.40

        pizza + biryani = 0.65  (stop here)

        Allowed tokens:[pizza, biryani]

2. Top-P = 0.80

        pizza = 0.40
        
        biryani = 0.65
        
        pasta = 0.80  (stop)

        Allowed tokens: [pizza, biryani, pasta]


⚙️ When both are set, the model applies them in order:

    1.Filter top-K tokens.

    2.Apply top-P probability cutoff.

    3.Sample from the remaining tokens using the temperature value.

🧠 Together, they balance creativity vs. control in the generated output.


```python
import os

from dotenv import load_dotenv

from google import genai

from google.genai import types
```

```python

load_dotenv()
```

```python

api_key = os.getenv("GOOGLE_API_KEY")
```

```python

client = genai.Client(api_key=api_key)
```

```python

# Generation configuration

config = types.GenerateContentConfig(
    temperature=0.7,
    top_p=0.9,
    top_k=40
)
```

```python
# Prompt

prompt = "Explain Gemini models in simple terms for beginners in 2 sentences."
```

```python

# Generate response
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt,
    config=config
)
```

```python

print(response.text)
```
