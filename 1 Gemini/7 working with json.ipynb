### Working with JSON

```python
from google import genai
import os
import json
from dotenv import load_dotenv

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GOOGLE_API_KEY")
)
```

```python
student = {
    "name": "Rahul",
    "age": 21,
    "course": "AI and Machine Learning",
    "skills": ["Python", "SQL", "Machine Learning"],
    "marks": {
        "Python": 85,
        "SQL": 78,
        "ML": 92
    }
}
```

```python

prompt = f"""
Analyze the following student data.

Student JSON:
{json.dumps(student, indent=2)}

Give:
1. Student name
2. Average marks
3. Best subject
4. A short performance comment
"""
```

```python

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```
