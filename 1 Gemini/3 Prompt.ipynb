
#### A prompt in AI is a piece of input given to an artificial intelligence (AI) model to produce a specific response. 


### Prompt engineering is the process of asking the right questions to get the best output from an AI model. 

#### It allows for direct interaction with the model using plain language prompts. (formatted strings)

```python
from dotenv import load_dotenv

import os

from google import genai
```

```python
# Load .env
load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
```

```python
# Create Gemini client

client = genai.Client(api_key=api_key)
```

```python
prompt = "Explain AI to me like I'm a kid."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

### Instructional Prompt

Tell the model to do something clearly.

An instructional prompt is when you tell the model exactly what task to do using a direct command or request.

    No examples (unlike few-shot).
    
    Just clear instructions.
    

```python
prompt = """
Summarize the following paragraph in one sentence:

Artificial intelligence (AI) is technology that enables computers and machines to simulate human learning, comprehension, problem solving, decision making, creativity and autonomy.

Applications and devices equipped with AI can see and identify objects. They can understand and respond to human language. They can learn from new information and experience. They can make detailed recommendations to users and experts. They can act independently, replacing the need for human intelligence or intervention (a classic example being a self-driving car).

"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

### Zero-shot Prompting

Zero-shot = 0 examples

Just ask directly without examples.

    The model is asked to perform a task without seeing any examples in the prompt.

    It relies only on its pre-trained knowledge.

    You just give the instruction, and the model figures it out.
    

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Translate 'How are you?' into Hindi."
)

print(response.text)
```

### One-Shot Prompting

One-shot = exactly 1 example

```python
prompt = """
Convert the sentence to past tense.

Example:
She goes to school. → She went to school.

Now:
He eats an apple.
"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

### Few-Shot Prompting

Few-shot = multiple examples

```python

prompt = """
Classify the sentiment.

Example 1:
"I love this phone." → Positive

Example 2:
"This product is terrible." → Negative

Example 3:
"The camera is excellent." → Positive

Now classify:
"The battery is disappointing."
"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

| Method        | Examples Given in Prompt | When Used                                       |
| ------------- | ------------------------ | ----------------------------------------------- |
| **Zero-Shot** | 0                        | When model is strong enough / task is simple    |
| **One-Shot**  | 1                        | When format/pattern matters, but task is simple |
| **Few-Shot**  | 2–10                     | When task is complex and model needs guidance   |


### Multi-turn Chat Prompt

Maintain conversational context.

Multi-turn chat prompting means giving the model a conversation history (previous user + assistant messages) so it can keep context across multiple turns, like in a real chat.

Instead of one isolated prompt, you pass the model a sequence of messages (role + content).

    system → sets rules / behavior
    
    user → what the human says
    
    assistant → what the model previously replied

This way, the model knows “what has already been said” and can continue smoothly.

```python

# Create a chat session
chat = client.chats.create(
    model="gemini-2.5-flash"
)

# First message
response1 = chat.send_message(
    "What is Python?"
)

print(response1.text)

# Second message
response2 = chat.send_message(
    "What is the difference between Python and Java?"
)

print(response2.text)
```

### Data Analysis Prompt

Ask the model to analyze or interpret data.

```python
prompt = """
You are a helpful data analyst.

Analyze this sales data:

Product, Region, Sales
A, North, 1200
B, North, 2300
A, South, 1100
C, North, 900
B, South, 1800
C, South, 700

Provide:

- Top-performing product and region
- Trends or patterns
- Suggestions to improve sales
"""

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```
