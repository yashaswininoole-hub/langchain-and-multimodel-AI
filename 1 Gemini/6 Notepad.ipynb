### Working with notepad doc

```python
from dotenv import load_dotenv

import os

from google import genai
```

```python
import sqlite3

import pandas as pd
```

```python

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
```

```python

client = genai.Client(api_key=api_key)
```

```python
def read_txt_file(file_path):

    try:
        with open(file_path, "r") as f:
            text = f.read()

        return text.strip()

    except Exception as e:
        print(f"Error reading file: {e}")
        return ""
```

```python

def summarize_text(text):

    prompt = f"""
You are a helpful ML Data Analyst.

Analyze the following text and extract information about:

1. Data Model Optimization Objective
2. Data Model Optimization Steps
3. Key Activities

Text:
{text}

Give the answer in simple language
using concise bullet points.
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python

txt_file = "ML_Lifecycle.txt"

txt_text = read_txt_file(txt_file)

if txt_text:

    summary = summarize_text(txt_text)

    print("FINAL SUMMARY")

    print(summary)
else:
    print("No text could be read from the file.")
```
