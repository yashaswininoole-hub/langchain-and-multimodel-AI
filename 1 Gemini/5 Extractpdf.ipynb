```python
from dotenv import load_dotenv

import os

from google import genai

from pypdf import PdfReader
```

```python
load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")

client = genai.Client(api_key=api_key)
```

```python

def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)

    text = ""

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n\n"

    return text
```

```python
def summarize_text(text, max_chars=10000):

    prompt = f"""
You are a helpful assistant.

Analyze the following PDF content:

{text[:max_chars]}

Extract information about the **Data Preparation** step.

Give the answer in the following format:

1. Objective
   - ...

2. Key Activities
   - ...
   - ...
   - ...

3. Step-by-Step Process
   - Step 1: ...
   - Step 2: ...
  
Keep the explanation simple and concise.
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python

pdf_file = "Life Cycle.pdf"
```

```python

pdf_text = extract_text_from_pdf(pdf_file)
```

```python
pdf_text
```

```python

summary = summarize_text(pdf_text)
```
