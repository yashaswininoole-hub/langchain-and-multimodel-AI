```python

from google import genai

client = genai.Client(api_key="")
```

```python

prompt = "Write about Data Science definition in one sentence ?"

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```


#### client is the object used to communicate with Gemini.
    

### client.models.generate_content() 

  * is the main method in the new Google GenAI SDK used to send a prompt to Gemini and get a complete response.

| Parameter  | Meaning                   |
| ---------- | ------------------------- |
| `model`    | Which Gemini model to use |
| `contents` | Your prompt/input         |


### create a fun to get data from LLM

```python

def gemini_response(prompt):
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    print("Prompt:", prompt)
    print("Response:\n", response.text)

    return response.text
```

```python

gemini_response("What's the Job roles of Gen AI LLMs in 3 sentenses?")
```

### Create a chat session to preserve history


```python

chat = client.chats.create(
    model="gemini-2.5-flash"
)

def chat_with_gemini(user_input):

    response = chat.send_message(user_input)

    print("User:", user_input)
    print("Gemini:", response.text)

    return response.text
```

```python
chat_with_gemini("My name is Ganapathi.")
```

```python

chat_with_gemini("What is my name?")
```

### Streaming response for single prompt

| Setting        | Behavior                            | Example                                      |
| -------------- | ----------------------------------- | -------------------------------------------- |
| `stream=False` | Wait for full output                | Best for backend scripts or batch processing |
| `stream=True`  | Receive words as they are generated | Best for chat apps or interactive UIs        |


stream=False → get complete model response after generation.

stream=True → get partial responses in real time (token streaming).

```python

def stream_gemini_response(prompt):

    response_stream = client.models.generate_content_stream(
        model="gemini-2.5-flash",
        contents=prompt
    )

    print("Gemini (streaming):", end=" ", flush=True)

    for chunk in response_stream:
        print(chunk.text, end="", flush=True)
```

```python

stream_gemini_response(
    "Explain how Large Language Models understand text in 3 sentence"
)
```

stream=True yields chunks as they arrive.

chunk.text is the partial text for that chunk.

### imp

```python

generate_content()        → complete response

generate_content_stream() → response chunk by chunk
```
