### DATA Loaders

### TextLoader


* TextLoader is used to load text files (.txt) into LangChain document format.

* LangChain stores documents as Document objects.

* Each document contains:

    * page_content → actual text

    * metadata → file information


```python

from langchain_community.document_loaders.text import TextLoader
```

```python

loader = TextLoader('speech.txt')

loader
```

 * At this stage file is not loaded yet.

```python

text_documents=loader.load()

text_documents
```


* Now LangChain:
    
   * Opens the file
    
   * Reads text
    
   * Converts it into Document objects
 
   * load() returns a list of documents.
 

| Field          | Meaning          |
| -------------- | ---------------- |
| `page_content` | actual text      |
| `metadata`     | file information |



```python

text = text_documents[0].page_content

print(text) 
```

### Loading from Multiple docs

```python

from langchain_community.document_loaders import TextLoader

files = ["ds.txt", "ml.txt"]

documents = []

for file in files:
    
    loader = TextLoader(file)
    
    docs = loader.load()
    
    documents.extend(docs)

print(len(documents))
```

```python

print(documents[0].page_content)
```

```python

print(documents[1].page_content)
```

### Using TextLoader inside DirectoryLoader

```python

from langchain_community.document_loaders import DirectoryLoader, TextLoader

loader = DirectoryLoader(
    "docs1",
    glob="*.txt",
    loader_cls=TextLoader
)

documents = loader.load()

print(len(documents))
```

```python

print(documents[0].page_content[:500])
```

```python

print(documents[1].page_content[:500])
```

### Access Each Document

```python

for doc in documents:
    
    print(doc.metadata["source"])
    print(doc.page_content)
```

### Read a PDf file


* It is used to read PDF files and convert them into LangChain Document objects.
  

```python

from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader('AI and DATA SCIENCE.pdf')
```

* At this stage:

   * The file is not yet read
    
   *  Only a loader object is created

```python

docs=loader.load()

docs
```

* Now LangChain:

    * Opens the PDF
    
    * Reads each page
    
    * Converts each page into a Document object

```python

print(len(docs))        # number of pages
```

```python

print(docs[0].page_content) 
```

```python

print(docs[0].metadata)
```

### Print Text from All Pages (Very Useful)

```python
for doc in docs:
    
    print("Page:", doc.metadata["page"] + 1)
    
    print(doc.page_content[:1000])
    
    print("-"*50)
```

### CSVLoader

* CSVLoader is a LangChain loader that reads a CSV file and converts each row into a Document object.

```python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(
    file_path="sales.csv"
)

documents = loader.load()

print("Total Rows:", len(documents))

for doc in documents:
    print(doc.page_content)
    print("-"*40)
```

### WebBaseLoader


For loading content from websites into LangChain, there are several web document loaders.

These fetch webpages and convert them into Document objects.

```python

from langchain_community.document_loaders import WebBaseLoader

url = "https://docs.langchain.com/oss/python/langchain/overview"

loader = WebBaseLoader(url)
docs = loader.load()

print("Total Documents:", len(docs))
print("\nContent Preview:\n")

print(docs[0].page_content[:2000])
```

```python
for i, doc in enumerate(docs):
    print(f"\nDocument {i+1}")
    print(doc.page_content[:1000])

    print("-"*50)
```
