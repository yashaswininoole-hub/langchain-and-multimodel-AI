### RecursiveCharacterTextSplitter


* The RecursiveCharacterTextSplitter is a document-splitting utility from LangChain.

   * Document-splitting utility from LangChain
   * Used to break large text into smaller chunks
   * Performs splitting recursively
   * Enables processing by LLMs
   * Allows storage in vector databases
   * Supports RAG (Retrieval Augmented Generation) systems



### Why RecursiveCharacterTextSplitter is used

* LLMs have token limits, so large documents must be split.

This splitter:

    * Breaks text into chunks of fixed size
    
    * Keeps semantic boundaries when possible
    
    * Uses multiple separators recursively
    
* Typical separators: ["\n\n", "\n", " ", ""]

### It tries splitting by:

    1 Paragraph
    2 Line
    3 Word
    4 Character

So the text structure is preserved as much as possible.

| Parameter         | Meaning                                    |
| ----------------- | ------------------------------------------ |
| `chunk_size`      | Maximum size of each chunk                 |
| `chunk_overlap`   | Overlap between chunks                     |
| `separators`      | Rules used to split text                   |
| `length_function` | Function to measure length (default = len) |


### PDF data

```python
import warnings
warnings.filterwarnings("ignore")
```

```python

from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader('AI and DATA SCIENCE.pdf')
```

```python

docs=loader.load()
```

```python

len(docs)
```

```python

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)

final_documents=text_splitter.split_documents(docs)

len(final_documents)
```
