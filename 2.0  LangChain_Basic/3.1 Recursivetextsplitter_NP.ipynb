### CharacterTextSplitter


* CharacterTextSplitter splits text into fixed-size chunks purely based on character count —

* it does not try to preserve sentence or paragraph boundaries like RecursiveCharacterTextSplitter does.

* What CharacterTextSplitter Does

      * Splits text based on a specific separator

      * Creates fixed-size chunks

      * Supports overlapping chunks

  

| Parameter         | Description                   |
| ----------------- | ----------------------------- |
| `separator`       | Character used to split text  |
| `chunk_size`      | Maximum characters in a chunk |
| `chunk_overlap`   | Overlap between chunks        |
| `length_function` | Function to measure length    |


```python

from langchain_text_splitters import CharacterTextSplitter

text_splitter=CharacterTextSplitter(separator="\n\n",chunk_size=100,chunk_overlap=20)

text_splitter.split_documents(docs)
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="",     
    chunk_size=100,
    chunk_overlap=20
)
docs = text_splitter.create_documents([speech])

for i, d in enumerate(docs[:5]):
    print(f"Chunk {i}: {len(d.page_content)} chars")
    print(d.page_content)
```

| Feature               | CharacterTextSplitter | RecursiveCharacterTextSplitter |
| --------------------- | --------------------- | ------------------------------ |
| Splitting logic       | Fixed separator       | Hierarchical, smart            |
| Meaning preservation  | ❌ Low                 | ✅ High                         |
| Chunk quality for RAG | ⚠️ Okay               | 🔥 Excellent                   |
| Speed                 | Fast                  | Slightly slower                |
| Best for              | Clean/structured text | PDFs, articles, docs, RAG      |
