
### Chroma is a vector database designed to store and search data using embeddings.

##### ChromaDB stores your documents + their embeddings + metadata, and helps you find documents similar to a user's query.


```python
# pip install chromadb
```

### Store and Search Simple Documents

```python

import chromadb
```

```python
# Create ChromaDB client

client = chromadb.Client()
```

```python
# Create collection

collection = client.create_collection( name="books")
```

```python

collection.add(
    ids=["1", "2", "3", "4"],
    documents=[
        "Python is a popular programming language.",
        "Machine learning uses data to learn patterns.",
        "ChromaDB is a vector database for AI applications.",
        "Deep learning uses neural networks."
    ]
)
```

```python
# Search

results = collection.query(
    query_texts=["What is used for AI applications?"],
    n_results=2
)
```

```python
print(results)
```

```python
print("Documents:")
print(results["documents"])

print("\nIDs:")
print(results["ids"])

print("\nDistances:")
print(results["distances"])
```

### Persistent Stored in cwd

```python
import chromadb

client = chromadb.PersistentClient(
    path="chroma_db"
)

collection = client.get_or_create_collection(
    name="books"
)
```

```python
collection.add(
    ids=["1", "2", "3", "4"],
    documents=[
        "Python is a programming language.",
        "Machine learning learns patterns from data.",
        "ChromaDB is a vector database for AI applications.",
        "Deep learning uses neural networks."
    ]
)
```

```python
results = collection.query(
    query_texts=["What is machine learning?"],
    n_results=2
)
```

```python
print("Results:")
print(results)

print("\nDocuments:")
print(results["documents"])

print("\nIDs:")
print(results["ids"])

print("\nDistances:")
print(results["distances"])
```

### Persistent data with meta data

```python
import chromadb

client = chromadb.PersistentClient(
    path="chroma_db"
)

collection = client.get_or_create_collection(
    name="knowledge1"
)

collection.add(
    ids=["1", "2", "3", "4"],

    documents=[
        "Python is a programming language.",
        "ChromaDB is a vector database.",
        "Soil pH should be between 6.5 and 7.5.",
        "Nitrogen is important for plant growth."
    ],

    metadatas=[
        {"category": "programming"},
        {"category": "AI"},
        {"category": "agriculture"},
        {"category": "agriculture"}
    ]
)
```

```python
results = collection.query(
    query_texts=["What is the ideal soil pH?"],
    n_results=2
)

print("Documents:")
print(results["documents"])

print("\nIDs:")
print(results["ids"])

print("\nMetadata:")
print(results["metadatas"])

print("\nDistances:")
print(results["distances"])
```

#### ChromaDB uses the l2 (Euclidean) distance metric by default for a newly created collection.

    It measures the straight-line distance between two embedding vectors.

| Metric             | ChromaDB name | Meaning                                  |
| ------------------ | ------------- | ---------------------------------------- |
| **L2 / Euclidean** | `"l2"`        | Straight-line distance                   |
| **Cosine**         | `"cosine"`    | Measures angle/direction between vectors |
| **Inner Product**  | `"ip"`        | Measures vector similarity               |
